A menu-driven program that stores student grades in a dictionary (student name as the key, grade as the value), using dictionaries, loops, functions and conditionals.

To keep it close to my own field, I have framed it as the grade book for a **Materials Science 101** class. Each of the six required functions is in its own cell below, and at the end there is a short bonus that applies the same dictionary operations to real materials data.

## Setup: the grades dictionary
This one dictionary is the whole grade book for the class. Every function reads from or writes to it. Keys are student names, values are their grades.

```python
# Materials Science 101 grade book: student name (key) -> grade (value)
grades = {}
```

## 1. add_grade()
Add a new student and their grade to the class grade book, then confirm.

*Example: adding Ali, who scored 85 in Materials Science 101.*

```python
def add_grade():
    # register a student and the grade they earned in the course
    name = input("Enter student name: ")
    grade = input("Enter grade: ")
    grades[name] = grade
    print(name + "'s grade added successfully!")
```

## 2. show_all_grades()
Print every student and grade in the class. If the grade book is empty, say so.

```python
def show_all_grades():
    if len(grades) == 0:
        print("No grades available.")
    else:
        print("--- All Grades ---")
        for name in grades:
            print(name + ": " + str(grades[name]))
```

## 3. show_specific_grade()
Look up one student. If they are in the grade book, print their grade; otherwise print an error.

```python
def show_specific_grade():
    name = input("Enter the student's name: ")
    if name in grades:
        print(name + "'s grade: " + str(grades[name]))
    else:
        print("Student not found.")
```

## 4. update_grade()
Correct a student's grade (for example after a re-mark). If the student exists, ask for the new grade and overwrite it; otherwise print an error.

```python
def update_grade():
    name = input("Enter the student's name to update: ")
    if name in grades:
        new_grade = input("Enter the new grade: ")
        grades[name] = new_grade
        print(name + "'s grade updated successfully!")
    else:
        print("Student not found.")
```

## 5. delete_grade()
Remove a student's record from the grade book. If they exist, delete the entry; otherwise print an error.

```python
def delete_grade():
    name = input("Enter the student's name to delete: ")
    if name in grades:
        del grades[name]
        print(name + "'s record deleted successfully!")
    else:
        print("Student not found.")
```

## 6. menu()
Show options 1 to 6, call the correct function for the chosen number, and keep looping until Exit is selected.

```python
def menu():
    while True:
        print()
        print("===== Grade Management System =====")
        print("1. Add Student Grade")
        print("2. Show All Grades")
        print("3. Show Specific Student's Grade")
        print("4. Update Grade")
        print("5. Delete Grade")
        print("6. Exit")

        choice = input("Choose an option (1-6): ")

        if choice == "1":
            add_grade()
        elif choice == "2":
            show_all_grades()
        elif choice == "3":
            show_specific_grade()
        elif choice == "4":
            update_grade()
        elif choice == "5":
            delete_grade()
        elif choice == "6":
            print("Exiting program. Goodbye!")
            break
        else:
            print("Invalid option. Please choose a number from 1 to 6.")
```

## Run the program
Run this cell and use the menu. Choose 6 to exit.

```python
menu()
```

## Bonus (materials science extension)

The same dictionary skills, applied to my own field. A grade book maps *student -> grade*; here I map *material -> Young's modulus* (a stiffness property, in gigapascals) and run the exact same operations: add, view all, view one, update and delete.

This version is scripted (no `input()`), so it runs top to bottom and prints its output. The values are approximate textbook figures for stiffness at room temperature.

```python
# material (key) -> Young's modulus in GPa (value), approximate textbook values
materials = {
    "Aluminium": 69,
    "Copper": 117,
    "Mild steel": 200,
    "Titanium": 116,
}

# 1. ADD a new material
materials["Tungsten"] = 411
print("Tungsten added successfully!")

# 2. VIEW ALL materials
print()
print("--- All Materials (Young's modulus, GPa) ---")
for name in materials:
    print(name + ": " + str(materials[name]))

# 3. VIEW a SPECIFIC material
target = "Titanium"
if target in materials:
    print(target + "'s stiffness: " + str(materials[target]) + " GPa")
else:
    print("Material not found.")

# 4. UPDATE a value (refine Copper's modulus)
if "Copper" in materials:
    materials["Copper"] = 110
    print("Copper's value updated successfully!")

# 5. DELETE a material
if "Mild steel" in materials:
    del materials["Mild steel"]
    print("Mild steel record deleted successfully!")

print()
print("--- Final materials list ---")
for name in materials:
    print(name + ": " + str(materials[name]) + " GPa")
```
